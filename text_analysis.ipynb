# Text Analysis
This notebook counts words, characters, vowels and repeated words in a text.

```python
text = ("Python is a popular language for data analysis. Python is easy to learn, and data analysis "
        "with Python is fun. Many students learn Python because Python is simple and powerful. "
        "Data helps us make better decisions, and good decisions need good data.")
```

```python
words_list = text.split()
word_count = len(words_list)
char_count = len(text)
char_no_space = len(text.replace(" ", ""))
```

```python
vowels = "aeiou"
vowel_count = 0
for ch in text.lower():
    if ch in vowels:
        vowel_count += 1
```

```python
import string
from collections import Counter

clean_text = text.lower().translate(str.maketrans("", "", string.punctuation))
words = clean_text.split()
word_freq = Counter(words)
repeated = {word: count for word, count in word_freq.items() if count > 1}
```

```python
print("Total words:", word_count)
print("Total characters (with spaces):", char_count)
print("Total characters (without spaces):", char_no_space)
print("Total vowels:", vowel_count)
print("Repeated words:")
for word, count in sorted(repeated.items(), key=lambda x: (-x[1], x[0])):
    print(f"  {word}: {count}")
```
